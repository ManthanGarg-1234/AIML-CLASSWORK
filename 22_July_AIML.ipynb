## 1. Creating Pandas Series

Pandas Series are one-dimensional labeled arrays capable of holding any data type (integers, strings, floats, Python objects, etc.). The axis labels are collectively referred to as the index.

### 1.1 From a List of Numbers

```python
import pandas as pd
data = ['a', 'b', 'c', 'd', 'e']
series = pd.Series(data)
print(series)
```

## 2. Creating Pandas DataFrames

Pandas DataFrames are two-dimensional labeled data structures with columns of potentially different types. You can think of them like a spreadsheet or SQL table, or a dictionary of Series objects.

### 2.1 From a Dictionary of Lists (Simple DataFrames)

## 5. DataFrame Selection with `.loc` and `.iloc`

Pandas provides powerful ways to select data using labels (`.loc`) or integer positions (`.iloc`).

Here, we'll demonstrate using a sample DataFrame created locally for clarity.

```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}
df = pd.DataFrame(data)
print(df)
print(df['Name'].min())
print(df['Name'].max())
print(df['Age'].min())
print(df['Age'].max())
print(df['Age'].mean())
```

### 5.1 Selecting Columns

```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}
df = pd.DataFrame(data)
print(df['Name'])
```

### 5.2 Selecting Rows and Columns by Label (`.loc`)

```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}
df = pd.DataFrame(data)
print(df.loc[2,'Name'])

print(df.loc[0:2,'Name']) # Name wali row ka 0 to 1 wala aa jayega
```

### 5.3 Selecting Rows and Columns by Integer Position (`.iloc`)

```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}
df = pd.DataFrame(data)

# print(df.iloc[2,2])
# print(df.iloc[1,0])
# print(df.iloc[0,1])

print(df.iloc[0:2,2]) # 2nd row ka 0 to 1 wala aa jayega
```

## 6. Conditional Selection / Filtering

Filter DataFrames based on conditions applied to one or more columns.

```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}
df = pd.DataFrame(data)

print(df[df['Age'] > 21])
```

## 7. Grouping and Aggregation

Group data based on one or more columns and then apply an aggregation function (like sum, mean, count, etc.) to the groups.

This example uses a local DataFrame for a clear demonstration of `groupby` functionality.

```python
# Grouping And Aggregation

import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}
df = pd.DataFrame(data)

# Group by 'City' and calculate the mean of 'Age' for each city
grouped_by_city_age_mean = df.groupby('City')['Age'].mean()
display(grouped_by_city_age_mean)

# You can also group by 'Age' and see descriptive statistics for other numeric columns if they exist
# For example, if there were another numeric column, say 'Score':
# df_with_score = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35], 'Score': [90, 85, 92]})
# grouped_by_age_score_mean = df_with_score.groupby('Age')['Score'].mean()
# display(grouped_by_age_score_mean)
```

```python
# Grouping And Aggregation

import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}
df = pd.DataFrame(data)

# Group by 'City' and calculate the mean of 'Age' for each city
grouped_by_city = df.groupby('City')
display(grouped_by_city.size())
```
