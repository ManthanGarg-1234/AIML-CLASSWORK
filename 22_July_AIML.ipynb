```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}
df = pd.DataFrame(data)
print(df)
print(df['Name'].min())
print(df['Name'].max())
print(df['Age'].min())
print(df['Age'].max())
print(df['Age'].mean())
```
