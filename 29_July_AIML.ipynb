```python
import pandas as pd
data1 = {
    'Department' :['IT','HR','IT','HR','IT','HR'],
    'Employee':['A','B','C','D','E','F'],
    'Salary':[50000, 60000, 55000, 62000, 58000, 61000]
}
data2 = {
    'Department' :['Sales','Finance','Sales','Finance','Sales','Finance'],
    'Employee':['G','H','I','J','K','L'],
    'Salary':[60000, 50000, 62000, 55000, 61000, 58000]
}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df_merged = pd.concat([df1,df2]).reset_index()
print(df_merged)
```

```python
import pandas as pd
data1 = {
    'Name':['A','B','C'],
    'Age':[20, 21, 22]
}
data2 = {
    'Name':['A','E','B'],
    'Age':[23, 24, 25]
}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
data3 = pd.merge(df1, df2, on='Name',how='left')
print(data3)
```
