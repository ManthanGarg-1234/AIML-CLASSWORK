### Theory: Pandas `groupby()`

The `groupby()` method is a fundamental function in Pandas for grouping rows together based on one or more column values. It allows you to split the data into groups, apply a function to each group independently, and then combine the results into a single DataFrame. This process is often referred to as 'Split-Apply-Combine'.

Key aspects:

*   **Splitting**: Data is divided into groups based on some criteria.
*   **Applying**: A function (e.g., aggregation, transformation, filtration) is applied to each group.
*   **Combining**: The results of applying the function are combined back into a new DataFrame or Series.

Common aggregate functions used with `groupby()` include:

*   `.mean()`: Calculates the mean of each group.
*   `.sum()`: Calculates the sum of each group.
*   `.count()`: Counts the number of non-null items in each group.
*   `.min()`/`.max()`: Finds the minimum/maximum value in each group.
*   `.first()`/`.last()`: Returns the first/last entry of each group.
*   `.size()`: Returns the size of each group.

```python
import pandas as pd
data1 = {
    'Department' :['IT','HR','IT','HR','IT','HR'],
    'Employee':['A','B','C','D','E','F'],
    'Salary':[50000, 60000, 55000, 62000, 58000, 61000]
}
data2 = {
    'Department' :['Sales','Finance','Sales','Finance','Sales','Finance'],
    'Employee':['G','H','I','J','K','L'],
    'Salary':[60000, 50000, 62000, 55000, 61000, 58000]
}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df_merged = pd.concat([df1,df2]).reset_index()
print(df_merged)
```

### Theory: Pandas `concat()`

The `pd.concat()` function is used to concatenate Pandas objects along a particular axis. This is primarily used to combine DataFrames or Series either by stacking them vertically (row-wise, `axis=0`) or horizontally (column-wise, `axis=1`).

Key parameters:

*   `objs`: A sequence or mapping of Series or DataFrame objects.
*   `axis`: The axis to concatenate along. `0` for row-wise (default), `1` for column-wise.
*   `join`: How to handle indexes on other axis(es). `inner` for intersection, `outer` for union (default).
*   `ignore_index`: If `True`, do not use the index values along the concatenation axis. The resulting axis will be labeled 0, ..., n-1. This is useful when you don't care about the original indices and want a fresh, continuous index, as shown in the example with `.reset_index()`.

```python
import pandas as pd
data1 = {
    'Department' :['IT','HR','IT','HR','IT','HR'],
    'Employee':['A','B','C','D','E','F'],
    'Salary':[50000, 60000, 55000, 62000, 58000, 61000]
}
data2 = {
    'Department' :['Sales','Finance','Sales','Finance','Sales','Finance'],
    'Employee':['G','H','I','J','K','L'],
    'Salary':[60000, 50000, 62000, 55000, 61000, 58000]
}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df_merged = pd.concat([df1,df2]).reset_index()
print(df_merged)
```

### Theory: Pandas `merge()`

The `pd.merge()` function is used to combine two DataFrames based on common columns or indices. It's similar to SQL `JOIN` operations and is highly flexible.

Key parameters:

*   `left`, `right`: The DataFrames to merge.
*   `on`: Column or list of column names to join on. These must be found in both DataFrames.
*   `how`: Type of merge to be performed. This specifies how to handle keys that appear in one DataFrame but not the other.

Types of merges (`how` parameter):

*   **`'inner'` (default)**: Uses intersection of keys from both DataFrames. It returns only the rows where the `on` key is present in *both* DataFrames. (Example: `XpFpOVaxXdpI`)
*   **`'left'`**: Uses keys from the left DataFrame only. It includes all rows from the left DataFrame and matching rows from the right DataFrame. If a key from the left DataFrame has no match in the right, the columns from the right DataFrame will have `NaN` values. (Example: `mcevo5dXZoOF`)
*   **`'right'`**: Uses keys from the right DataFrame only. It includes all rows from the right DataFrame and matching rows from the left DataFrame. If a key from the right DataFrame has no match in the left, the columns from the left DataFrame will have `NaN` values. (Example: `_aSyV0GVZsix`)
*   **`'outer'`**: Uses union of keys from both DataFrames. It includes all rows from both DataFrames, filling `NaN` for missing values in either DataFrame.
