```python
import pandas as pd

data = pd.read_csv("student.csv")

print(data.head())
print()
x = data.drop("Marks", axis = 1) # features

y = data["Marks"]  # target

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)

print("Training Data")
print()
print(x_train)
print()
print(y_train)
print()
print("Testing Data")
print()
print(x_test)
print()
print(y_test)


# 2nd part

from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(x_train, y_train)

print("Intercept:", model.intercept_)
print("\nCoefficient:", model.coef_)

y_pred = model.predict(x_test)

print("\nPredicted Values:", y_pred)
print("\nActual Values:", y_test)

# 3rd part

from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import root_mean_squared_error
from sklearn.metrics import r2_score

print("MAE = ",mean_absolute_error(y_test, y_pred))
print("MSE = ",mean_squared_error(y_test, y_pred))
print("RMSE = ",root_mean_squared_error(y_test, y_pred))
print("R2 = ",r2_score(y_test, y_pred))
```

### Visualizing the Regression Line

Let's visualize the relationship between 'Study hours' and 'Marks' and overlay the regression line predicted by our model. This will help us understand how well the model captures the underlying trend in the data.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming 'Study hours' is the feature used for prediction in this context
# We need to reshape x_test for plotting against y_test and y_pred
# For simplicity, let's plot against the 'Study hours' feature if it's the primary independent variable.
# If 'x' contains multiple features, we might need to choose one for 2D visualization.

# Let's assume 'Study hours' is at index 1 of the features if 'x' has 'Age' and 'Study hours'
# Or directly use x_test['Study hours'] if it's a DataFrame

plt.figure(figsize=(10, 6))
sns.scatterplot(x=x_test['Study hours'], y=y_test, label='Actual Marks', color='blue')
sns.lineplot(x=x_test['Study hours'], y=y_pred, label='Predicted Marks', color='red', linestyle='--')

plt.title('Actual vs. Predicted Marks based on Study Hours')
plt.xlabel('Study hours')
plt.ylabel('Marks')
plt.legend()
plt.grid(True)
plt.show()
```

### Model Evaluation Metrics Explained

*   **Mean Absolute Error (MAE):** This is the average of the absolute differences between predictions and actual values. It measures the average magnitude of the errors in a set of predictions, without considering their direction. A lower MAE indicates a better model.

*   **Mean Squared Error (MSE):** This is the average of the squared differences between predictions and actual values. It gives a higher penalty to larger errors due to the squaring. Like MAE, a lower MSE indicates a better model.

*   **R-squared (R2 Score):** This metric represents the proportion of the variance in the dependent variable that is predictable from the independent variables. It ranges from 0 to 1, where 1 indicates that the model explains all the variability of the response data around its mean, and 0 indicates no linear relationship. A higher R2 score indicates a better fit for the model.
