```python
from numpy._core.defchararray import center
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
X = np.array([
    [1,2],
    [1.5,1.8],
    [5,8],
    [8,8],
    [1,0.6],
    [9,11]
])
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)
kmeans.fit(X)
labels = kmeans.labels_
centers = kmeans.cluster_centers_
print("Cluster Lables: ")
print(labels)
print("\nCluster Centers: ")
print(centers)
plt.figure(figsize=(8,6))
plt.scatter(X[:,0],X[:,1],c=labels,cmap='viridis',s=100)
plt.scatter(centers[:,0],centers[:,1],marker="X",s=250,edgecolors="black",label="Centroids")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-means Clustering")
plt.legend()
plt.grid(True)
plt.show()
```

# Notes on K-means Clustering Experiment

### 1. Dataset Characteristics
- **Total Data Points**: 6 two-dimensional coordinates.
- **Data Coordinates**:
  - $[1, 2]$
  - $[1.5, 1.8]$
  - $[5, 8]$
  - $[8, 8]$
  - $[1, 0.6]$
  - $[9, 11]$

### 2. Algorithm Configuration
- **Algorithm**: K-means Clustering ($K=3$)
- **Parameters**:
  - `n_clusters=3`
  - `random_state=42`
  - `n_init=10`

### 3. Clustering Results & Analysis
- **Cluster Labels**: `[1, 1, 2, 2, 1, 0]`
- **Cluster Assignments & Centroids**:
  - **Cluster 0** (Centroid: $[9.0, 11.0]$): Contains a single outlier point at $[9, 11]$. Due to the low value of $K$ and sparse data structure, this point got isolated.
  - **Cluster 1** (Centroid: $[1.17, 1.47]$): Groups the compact cluster on the bottom-left ($[1, 2]$, $[1.5, 1.8]$, and $[1, 0.6]$).
  - **Cluster 2** (Centroid: $[6.5, 8.0]$): Groups the two points in the middle-right ($[5, 8]$ and $[8, 8]$).

### 4. Observations
- K-means successfully captured the high-density cluster at the bottom-left.
- Selecting $K=3$ for only 6 data points highlights how the algorithm handles isolated points (outliers) by assigning them to unique clusters.
